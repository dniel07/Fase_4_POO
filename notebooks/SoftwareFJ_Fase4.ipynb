## 1. Clientes y validación
La clase Cliente hereda de Entidad y protege sus atributos mediante propiedades.

```python
from src.clientes import Cliente
from src.excepciones import SoftwareFJError
cliente = Cliente('C100', 'Ana Torres', 'ana@example.com', '3001234567')
print(cliente.describir())
try:
    Cliente('C101', '', 'correo-malo', '123')
except SoftwareFJError as error:
    print('Error controlado:', error)
```

## 2. Servicios y polimorfismo
Los tres servicios implementan los mismos métodos, pero validan y describen su unidad de cobro.

```python
from src.servicios import ReservaSala, AlquilerEquipo, AsesoriaEspecializada
servicios = [ReservaSala('S100', 'Sala de capacitación', 50000), AlquilerEquipo('S101', 'Alquiler de portátil', 80000), AsesoriaEspecializada('S102', 'Asesoría técnica', 120000)]
for servicio in servicios: print(servicio.describir(), servicio.calcular_costo())
print('Equipo 3 días con impuesto y descuento:', servicios[1].calcular_costo(3, 0.19, 0.05))
```

## 3. Reservas y estados
Una reserva debe confirmarse antes de procesarse. Las operaciones no permitidas generan excepciones.

```python
from src.reservas import Reserva
reserva = Reserva('R100', cliente, servicios[0], 2, 0.19, 0.10)
reserva.confirmar()
print('Costo procesado:', reserva.procesar())
try:
    reserva.cancelar()
except SoftwareFJError as error:
    print('Operación bloqueada:', error)
```
